```python
# First, ensure your 'date' column is recognized as a datetime format by pandas
# (If your date is in format like "20231201" or "2023-12-01")
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# Now, create the new columns by extracting from the date
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day

# Finally, look at the perfectly standardized dates
print("\nSample of Standardized Dates:")
df[['date', 'year', 'month', 'day']].head()
# Notice the capital 'I' in 'Int64'
df['year'] = df['date'].dt.year.astype('Int64')
df['month'] = df['date'].dt.month.astype('Int64')
df['day'] = df['date'].dt.day.astype('Int64')

# Look at the perfectly standardized dates
print("\nSample of Standardized Dates:")
df[['date', 'year', 'month', 'day']].head()

```

```python
# ==========================================
# 10. Fix Duplicates and Negative Numbers
# ==========================================

# 1. Drop all duplicate rows
df = df.drop_duplicates()

# 2. Fix all impossible negative numbers (change them to 0)
# We select only the columns that actually have numbers (integers or floats)
numeric_cols = df.select_dtypes(include=['number']).columns

# Replace any value less than 0 with 0
df[numeric_cols] = df[numeric_cols].clip(lower=0)


print(f"✅ Duplicates dropped. New shape: {df.shape[0]:,} Rows")
print("✅ All negative values have been cleanly converted to 0.")
```

```python
# List of all the numeric columns that hold counts/people
numeric_cols = [
    'bio_age_5_17', 'bio_age_17_', 
    'demo_age_5_17', 'demo_age_17_', 
    'age_0_5', 'age_5_17', 'age_18_greater'
]

# Replace all missing (NaN) values with 0
df[numeric_cols] = df[numeric_cols].fillna(0)
```

```python
# Drop rows where 'date' is missing
df = df.dropna(subset=['date'])

# Check everything again
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()
```

```python
# ==========================================
# 9️⃣ Final Data Quality & Sanity Check
# ==========================================
print("=== FINAL DATA QUALITY REPORT ===\n")

# 1. Check for any remaining Missing Values
missing_vals = df.isnull().sum()
missing_cols = missing_vals[missing_vals > 0]
if missing_cols.empty:
    print("✅ PASS: No missing values (Null/NaN) found in the entire dataset!")
else:
    print("❌ WARNING: Missing values found in these columns:")
    print(missing_cols)

# 2. Check for exact duplicate rows
duplicates = df.duplicated().sum()
if duplicates == 0:
    print("\n✅ PASS: No duplicate rows exist.")
else:
    print(f"\n❌ WARNING: Found {duplicates:,} duplicate rows!")

# 3. Validate the State count
num_states = df['state'].nunique()
if num_states == 36:
    print(f"\n✅ PASS: Exactly 36 valid States/Union Territories found.")
else:
    print(f"\n❌ WARNING: Expected 36 states, but found {num_states}.")

# 4. Check Date format validation (Ensures they are converted to Datetime properly)
if pd.api.types.is_datetime64_any_dtype(df['date']):
    print("\n✅ PASS: The 'date' column is correctly formatted as a Datetime object.")
else:
    print("\n❌ WARNING: The 'date' column is still raw text (Object). You should convert it to Datetime.")

# 5. Check Pincodes (must be exactly 6 digits)
df['pincode_str'] = df['pincode'].astype(str)
invalid_pincodes = df[~df['pincode_str'].str.match(r'^\d{6}$')].shape[0]
if invalid_pincodes == 0:
    print("\n✅ PASS: All Pincodes are exactly 6 digits long.")
else:
    print(f"\n❌ WARNING: Found {invalid_pincodes:,} invalid pincodes (not 6 digits).")
df = df.drop(columns=['pincode_str']) # clean up temporary column

# 6. Check for negative numbers in Population/Age (Impossible values)
numeric_cols = df.select_dtypes(include=['number']).columns
negative_vals = (df[numeric_cols] < 0).sum().sum()
if negative_vals == 0:
    print("\n✅ PASS: No negative numbers found in numeric columns (like age or population).")
else:
    print(f"\n❌ WARNING: Found {negative_vals:,} instances of negative numbers!")

print("\n=================================")
print(f"Final Clean Dataset Shape: {df.shape[0]:,} Rows | {df.shape[1]} Columns")
print("=================================")
```
