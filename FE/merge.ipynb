```python
import pandas as pd

# ================================
# 1️⃣ Setup Paths for Cleaned Data
# ================================
bio_path   = r"C:\Users\Bala murukan\Desktop\Aadhaar_Insights\Aadhaar_Insight\FE\biometric\biometric_cleaned.csv"
demo_path  = r"C:\Users\Bala murukan\Desktop\Aadhaar_Insights\Aadhaar_Insight\FE\demographic\demographic_cleaned.csv"
enrol_path = r"C:\Users\Bala murukan\Desktop\Aadhaar_Insights\Aadhaar_Insight\FE\Enrolment\enrolment_cleaned.csv"

# Output path for the final merged file
output_path = r"C:\Users\Bala murukan\Desktop\Aadhaar_Insights\Aadhaar_Insight\FE\Master_Cleaned_Dataset.csv"

# ================================
# 2️⃣ Load Cleaned Datasets
# ================================
print("Loading files...")
df_bio   = pd.read_csv(bio_path)
df_demo  = pd.read_csv(demo_path)
df_enrol = pd.read_csv(enrol_path)

# ================================
# 3️⃣ Merge Datasets
# ================================
# The common columns shared across all three datasets to link them together
common_keys = ['date', 'state', 'district', 'pincode']

print("Merging Biometric and Demographic...")
merged_1 = pd.merge(df_bio, df_demo, on=common_keys, how='outer')

print("Merging with Enrolment...")
master_df = pd.merge(merged_1, df_enrol, on=common_keys, how='outer')

print("\n=== MASTER DATASET ===")
print(f"Rows    : {master_df.shape[0]:,}")
print(f"Columns : {master_df.shape[1]}")

# ================================
# 4️⃣ Save the Master Dataset
# ================================
print("\nSaving Master Dataset...")
master_df.to_csv(output_path, index=False)

print(f"\n✅ All files successfully merged and saved to {output_path}")
```
