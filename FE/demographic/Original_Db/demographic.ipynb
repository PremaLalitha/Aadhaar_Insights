```python
import pandas as pd

# ================================
# 1️⃣ Setup Paths
# ================================
input_path  = r"C:\Users\Bala murukan\Desktop\Aadhaar_Insights\Aadhaar_Insight\FE\demographic\Original_Db"
output_path = r"C:\Users\Bala murukan\Desktop\Aadhaar_Insights\Aadhaar_Insight\FE\demographic"

# ================================
# 2️⃣ Load Datasets (Notice there are 5 files here)
# ================================
df1 = pd.read_csv(f"{input_path}\\Demographic 1.csv")
df2 = pd.read_csv(f"{input_path}\\Demographic 2.csv")
df3 = pd.read_csv(f"{input_path}\\Demographic 3.csv")
df4 = pd.read_csv(f"{input_path}\\Demographic 4.csv")
df5 = pd.read_csv(f"{input_path}\\Demographic 5.csv")

# ================================
# 3️⃣ Merge Datasets
# ================================
merged_df = pd.concat([df1, df2, df3, df4, df5], ignore_index=True)

print("\n=== MERGED DATASET ===")
print(f"Rows    : {merged_df.shape[0]:,}")
print(f"Columns : {merged_df.shape[1]}")
print(f"Full    : {merged_df.shape}")

# ================================
# 4️⃣ Check Missing Values
# ================================
print("\n=== MISSING VALUES ===")
print(merged_df.isnull().sum())

# ================================
# 5️⃣ Check Duplicate Rows
# ================================
dup_before = merged_df.duplicated().sum()

print("\n=== DUPLICATE CHECK ===")
print(f"Duplicate rows : {dup_before:,}")

# ================================
# 6️⃣ Remove Duplicates
# ================================
clean_df = merged_df.drop_duplicates()

dup_after = clean_df.duplicated().sum()

print("\n=== AFTER CLEANING ===")
print(f"Final rows     : {clean_df.shape[0]:,}")
print(f"Duplicates now : {dup_after}")
print(f"Rows removed   : {dup_before:,}")

# ================================
# 7️⃣ Save Cleaned Dataset
# ================================
output_file = f"{output_path}\\demographic_cleaned.csv"

clean_df.to_csv(output_file, index=False)

print("\n✅ demographic_cleaned.csv saved successfully!")
```
